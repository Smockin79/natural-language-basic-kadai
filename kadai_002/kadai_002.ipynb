```python
from urllib.request import Request, urlopen
from bs4 import BeautifulSoup

url = "https://www.aozora.gr.jp/cards/000148/files/2371_13943.html"

request = Request(
    url,
    headers={"User-Agent": "Mozilla/5.0"}
)

with urlopen(request, timeout=20) as response:
    html = response.read().decode("shift_jis")

soup = BeautifulSoup(html, "html.parser")

print("ページを取得できました")
```

```python
main_text = soup.select_one("div.main_text")

if main_text is None:
    raise ValueError("本文が見つかりませんでした")

print(main_text.get_text()[:200])
```

```python
import re

# 漢字に付いている読み仮名を取り除く
for tag in main_text.select("rt, rp"):
    tag.decompose()

# HTMLタグを取り除いて文字だけにする
raw_text = main_text.get_text()

# 改行、空白、全角スペースを取り除く
clean_text = re.sub(r"[\s\u3000]+", "", raw_text)

print(clean_text[:300])
```

```python
first_sentence = clean_text.split("。", 1)[0] + "。"

print(first_sentence)
```

```python
words = [
    ("近頃", "名詞"),
    ("は", "助詞"),
    ("大分", "副詞"),
    ("方々", "名詞"),
    ("の", "助詞"),
    ("雑誌", "名詞"),
    ("から", "助詞"),
    ("談話", "名詞"),
    ("を", "助詞"),
    ("しろしろ", "動詞"),
    ("と", "助詞"),
    ("責められて", "動詞"),
    ("、", "記号"),
    ("頭", "名詞"),
    ("が", "助詞"),
    ("がらん胴", "名詞"),
    ("に", "助詞"),
    ("なった", "動詞"),
    ("から", "助詞"),
    ("、", "記号"),
    ("当分", "副詞"),
    ("品切れ", "名詞"),
    ("の", "助詞"),
    ("看板", "名詞"),
    ("でも", "助詞"),
    ("懸けたい", "動詞"),
    ("くらい", "助詞"),
    ("に", "助詞"),
    ("思っています", "動詞"),
    ("。", "記号"),
]

print(words)
```

```python
stop_pos = {"助詞", "記号"}

result = [
    word
    for word, pos in words
    if pos not in stop_pos
]

print("元の一文：")
print(first_sentence)

print("\nストップワード除去後：")
print(" / ".join(result))
```
